```python
import pandas as pd
import numpy as np
data={
    'E-name':['Roshan','Amar','Ashwini','Lohith','Mohan','Pramodh'],
    'Department':['CS','CS','EC','EC','CS','EC'],
    'Experience':[10,20,15,14,9,8],
    'Salary':[45000,50000,30000,25000.10000,40000]
}
df=pd.DataFrame(data)
print(df)
table=pd.pivot_table(df,index=['E-namee'Department'],
values='salary',aggfunc=['mean'])
print(table)
table1=pd.pivote_table(df,index=['E-name','Department'],
values='salary',aggfunc=['sum','mean'])
print(table1)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data={
    'number':[1,2,3,4,5],
    'pencil':[300,350,400,500,520],
    'textbook':[250,350,400,420,500],
    'drawingsheet':[100,200,200,250,200],
    'total':[800,1000,1320,1510,2000],
    'profit':[8000,9500,10256,1200,18000],
    }
df=pd.DataFrame(data)
print(df)
df.info()
plt.plot(df['number'],df['profit'])
plt.xlabel('number')
plt.ylabel('total profit')
plt.show()
print("missingvalue")
print(df.isnull().sum())
print("sum of the total profit")
print(df['profit'].sum())
print("max value of drawingsheet")
print(df["drawingsheet"].max())
print("min value of the textbook cloumn")
print(df["textbook"].min())
new_total=df["total"]+5
print(new_total)
```
