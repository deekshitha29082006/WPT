```python

# ============================================================
# YOGA POSE RECOGNITION USING CNN
# Training + Final Prediction + Streamlit App
# Google Colab | No ngrok
# ============================================================

import os
import sys
import json
import time
import re
import subprocess
import urllib.request
from pathlib import Path

# 1. Install required packages
subprocess.run([
    sys.executable, "-m", "pip", "install", "-q",
    "tensorflow", "streamlit", "pillow"
], check=True)

import numpy as np
import tensorflow as tf
from PIL import Image
from google.colab import files
from IPython.display import display

print("YOGA POSE RECOGNITION USING CNN")
print("TensorFlow version:", tf.__version__)

# 2. Create dataset folders
BASE_DIR = Path("/content/yoga_dataset")
POSES = ["MountainPose", "TreePose", "WarriorPose"]

for pose in POSES:
    (BASE_DIR / pose).mkdir(parents=True, exist_ok=True)

# 3. Upload images for each yoga pose
def upload_pose_images(pose):
    folder = BASE_DIR / pose
    print(f"\nUpload images for {pose}")
    print("Select multiple JPG, JPEG, or PNG images.")

    uploaded = files.upload()
    saved = 0

    for filename in uploaded:
        try:
            with Image.open(filename) as img:
                img = img.convert("RGB")
                output_name = f"image_{len(list(folder.glob('*.jpg'))):05d}.jpg"
                img.save(folder / output_name, "JPEG")
                saved += 1
        except Exception as e:
            print("Skipped invalid image:", filename, e)

    print(f"{pose}: saved {saved} images")

for pose in POSES:
    upload_pose_images(pose)

# 4. Check dataset
print("\nDATASET SUMMARY")

for pose in POSES:
    count = len(list((BASE_DIR / pose).glob("*.jpg")))
    print(pose, "=", count, "images")

    if count < 5:
        raise ValueError(
            f"Upload at least 5 valid images for {pose}. "
            "Run the cell again after adding more images."
        )

# 5. Load and split images
IMG_SIZE = (128, 128)
BATCH_SIZE = 8
SEED = 42

train_ds = tf.keras.utils.image_dataset_from_directory(
    str(BASE_DIR),
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="int",
    validation_split=0.2,
    subset="training",
    seed=SEED,
    class_names=POSES
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    str(BASE_DIR),
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="int",
    validation_split=0.2,
    subset="validation",
    seed=SEED,
    class_names=POSES
)

class_names = train_ds.class_names
print("\nClass names:", class_names)

# 6. Normalize pixel values
def normalize_images(images, labels):
    images = tf.cast(images, tf.float32) / 255.0
    return images, labels

train_ds = train_ds.map(normalize_images).prefetch(
    tf.data.AUTOTUNE
)
val_ds = val_ds.map(normalize_images).prefetch(
    tf.data.AUTOTUNE
)

# 7. Build CNN model
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(128, 128, 3)),
    tf.keras.layers.Conv2D(32, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(64, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(128, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dense(128, activation="relu"),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(len(POSES), activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# 8. Train CNN
print("\nTRAINING CNN MODEL")

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10
)

# 9. Save model and class names
MODEL_PATH = "/content/yoga_pose_model.keras"
CLASS_PATH = "/content/yoga_class_names.json"

model.save(MODEL_PATH)

with open(CLASS_PATH, "w") as f:
    json.dump(class_names, f)

print("\nCNN TRAINING COMPLETED")
print("Model saved:", MODEL_PATH)
print("Class names saved:", CLASS_PATH)

# 10. Final prediction
print("\nUPLOAD A NEW YOGA POSE IMAGE FOR PREDICTION")
test_uploads = files.upload()

for filename in test_uploads:
    try:
        img = Image.open(filename).convert("RGB")
        display(img)

        img = img.resize(IMG_SIZE)
        arr = np.asarray(img, dtype=np.float32) / 255.0
        arr = np.expand_dims(arr, axis=0)

        probabilities = model.predict(arr, verbose=0)[0]
        index = int(np.argmax(probabilities))
        predicted_pose = class_names[index]
        confidence = float(probabilities[index]) * 100

        print("\n========== FINAL PREDICTION ==========")
        print("Image:", filename)
        print("Predicted Yoga Pose:", predicted_pose)
        print(f"Model Confidence: {confidence:.2f}%")
        print("======================================")

    except Exception as e:
        print("Could not predict", filename, ":", e)

# 11. Create Streamlit application
APP_CODE = r'''
import json
import numpy as np
import streamlit as st
import tensorflow as tf
from PIL import Image

st.set_page_config(
    page_title="Yoga Pose Recognition",
    page_icon="🧘",
    layout="centered"
)

@st.cache_resource
def load_model():
    return tf.keras.models.load_model(
        "/content/yoga_pose_model.keras"
    )

model = load_model()

with open("/content/yoga_class_names.json", "r") as f:
    class_names = json.load(f)

st.title("🧘 Yoga Pose Recognition")
st.write("CNN-based Yoga Image Classification")
st.info("Upload an image of one of the trained yoga poses.")

uploaded = st.file_uploader(
    "Upload a yoga image",
    type=["jpg", "jpeg", "png"]
)

if uploaded:
    image = Image.open(uploaded).convert("RGB")
    st.image(image, caption="Uploaded Yoga Image",
             use_container_width=True)

    if st.button("Predict Yoga Pose"):
        image = image.resize((128, 128))
        arr = np.asarray(image, dtype=np.float32) / 255.0
        arr = np.expand_dims(arr, axis=0)

        probabilities = model.predict(arr, verbose=0)[0]
        index = int(np.argmax(probabilities))

        st.subheader("Final Prediction")
        st.success(f"Yoga Pose: {class_names[index]}")
        st.metric(
            "Model confidence",
            f"{probabilities[index] * 100:.2f}%"
        )
        st.caption(
            "Educational image classification only. "
            "The result does not assess posture safety or correctness."
        )
'''

with open("/content/app.py", "w") as f:
    f.write(APP_CODE)

print("\nStreamlit app created: /content/app.py")

# 12. Start Streamlit
with open("/content/streamlit.log", "w") as log:
    app_process = subprocess.Popen([
        sys.executable, "-m", "streamlit", "run",
        "/content/app.py",
        "--server.port", "8501",
        "--server.address", "0.0.0.0",
        "--server.headless", "true"
    ], stdout=log, stderr=subprocess.STDOUT)

# 13. Create a temporary public link using Cloudflare
try:
    tunnel_path = "/content/cloudflared"

    urllib.request.urlretrieve(
        "https://github.com/cloudflare/cloudflared/releases/latest/"
        "download/cloudflared-linux-amd64",
        tunnel_path
    )
    os.chmod(tunnel_path, 0o755)

    with open("/content/cloudflared.log", "w") as log:
        tunnel_process = subprocess.Popen([
            tunnel_path, "tunnel",
            "--url", "http://127.0.0.1:8501",
            "--no-autoupdate"
        ], stdout=log, stderr=subprocess.STDOUT)

    print("\nStarting the web application...")
    public_url = None

    for _ in range(45):
        time.sleep(2)

        with open("/content/cloudflared.log", "r") as f:
            log_text = f.read()

        match = re.search(
            r"https://[a-zA-Z0-9-]+\.trycloudflare\.com",
            log_text
        )

        if match:
            public_url = match.group(0)
            break

    if public_url:
        print("\nYOUR YOGA POSE APP LINK:")
        print(public_url)
        print("Keep the Colab session running.")
    else:
        print("Public link not found.")
        print("Check /content/cloudflared.log")
        print("Check /content/streamlit.log")

except Exception as e:
    print("\nCould not create public link:", e)
    print("CNN training and notebook prediction are complete.")
```
